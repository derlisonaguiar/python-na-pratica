## Conceito

Variáveis associam nomes a objetos. Tipos como `int`, `float`, `str` e `bool` ajudam a representar diferentes informações. Operadores permitem transformar dados.

```python
nome = "Ada"
idade = 28
nota = 9.5
print(f"{nome}: {idade} anos, nota {nota}")
print(type(idade), 10 // 3, 10 % 3)
```
