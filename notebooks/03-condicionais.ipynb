## Conceito

Condições escolhem caminhos com base em expressões booleanas. Use `if`, `elif` e `else`; combine comparações com `and`, `or` e `not`.

```python
nota = 7.5
frequencia = 85
if nota >= 7 and frequencia >= 75:
    print("Aprovado")
elif nota >= 5 and frequencia >= 75:
    print("Recuperação")
else:
    print("Reprovado")
```
